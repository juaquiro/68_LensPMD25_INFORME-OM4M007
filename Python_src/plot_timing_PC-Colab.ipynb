```python
# timing_barplot.ipynb
# =====================
# This notebook loads timing data from an Excel file and plots a comparison chart
# the timing data are compiled in oneNote->68 LENSPMD25 INFORME-OM4M007 sptial freq using ML->Comparacion CPUs vs GPUs solucion python + google colab
# -> CPU vs GPU



# --- 1. Imports ---
import pandas as pd
import matplotlib.pyplot as plt
import os

# --- 2. Load Excel file ---
# Adjust the path as needed
data_path = os.path.join("..", "local_data", "timing_comparison.xlsx")

df = pd.read_excel(data_path)

# Display the loaded data
print("Data loaded from:", data_path)
display(df)

# --- 3. Plot configuration ---
plt.figure(figsize=(10, 6))

# Set the metric column as x-axis
x = df["Metric"]

# Plot each column (skip the first one)
plt.bar(x, df["PC Numpy (CPU sec)"], width=0.15, label="PC Numpy (CPU)", align='center')
plt.bar([i + 0.15 for i in range(len(x))], df["PC Tensorflow (GPU sec)"], width=0.15, label="PC Tensorflow (GPU)")
plt.bar([i + 0.30 for i in range(len(x))], df["Colab T4 GPU Numpy (CPU sec)"], width=0.15, label="Colab T4 GPU Numpy (CPU)")
plt.bar([i + 0.45 for i in range(len(x))], df["Colab T4 GPU Tensorflow (GPU sec)"], width=0.15, label="Colab T4 GPU Tensorflow (GPU)")

# --- 4. Beautify plot ---
plt.yscale("log")  # ← Logarithmic vertical axis
plt.xticks([i + 0.25 for i in range(len(x))], x)
plt.xlabel("Operation Step")
plt.ylabel("Execution Time (seconds)")
plt.title("Timing Comparison Across Platforms and Libraries")
plt.legend()
plt.grid(axis="y", linestyle="--", alpha=0.7)

# --- 5. Show ---
plt.tight_layout()
plt.show()
```
